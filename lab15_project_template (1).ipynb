## Part 1: Dataset Overview และ EDA เบื้องต้น

ในส่วนนี้ให้โหลด dataset, แสดง basic statistics, และทำ EDA เบื้องต้นเพื่อเข้าใจข้อมูลก่อนวิเคราะห์

**สิ่งที่ต้องทำ:**
- โหลด dataset และแสดง shape, dtypes, missing values
- แสดง descriptive statistics ด้วย `.describe()`
- Plot distribution ของ target variable
- Plot correlation heatmap หรือ scatter matrix

```python
# 1. โหลดข้อมูล
df = pd.read_csv('data/WineQT.csv') 

# 2. ตัดคอลัมน์ Id ออก (เนื่องจากเป็นเพียง Index ไม่ใช่ตัวแปรทางเคมี)
if 'Id' in df.columns:
    df = df.drop(columns=['Id'])

print("=== ขนาดของ Dataset ===")
print(f"จำนวนแถว (Rows): {df.shape[0]}, จำนวนคอลัมน์ (Columns): {df.shape[1]}\n")

print("=== ตรวจสอบชนิดข้อมูลและ Missing Values ===")
df_info = pd.DataFrame({
    'Data Type': df.dtypes,
    'Missing Values': df.isnull().sum(),
    'Missing (%)': (df.isnull().sum() / len(df)) * 100,
    'Unique Values': df.nunique()
})
print(df_info)

print("\n=== ตรวจสอบแถวที่ซ้ำกัน (Duplicates) ===")
print(f"จำนวนแถวซ้ำ: {df.duplicated().sum()} แถว")

# แสดงสถิติพื้นฐาน (Descriptive Statistics)
df.describe().T
```

```python
# ─── Descriptive Statistics + Visualization ───────────────────────
# วัตถุประสงค์: ดูภาพรวมของข้อมูล

# 1. Summary statistics (df.describe)
print("=== สถิติเชิงพรรณนา (Descriptive Statistics) ===")
display(df.describe().T) # ใช้ display() แทน print() เพื่อให้ตารางใน VS Code ดูอ่านง่ายขึ้น

# ตั้งค่าขนาดพื้นที่กราฟ (กว้าง 15, สูง 5) สำหรับพล็อต 2 รูปคู่กัน
plt.figure(figsize=(15, 5))

# 2. Distribution plot ของ target ('quality')
plt.subplot(1, 2, 1)
sns.countplot(x='quality', data=df, palette='viridis', hue='quality', legend=False)
plt.title('Distribution of Wine Quality (Target Variable)', fontweight='bold')
plt.xlabel('Quality Score')
plt.ylabel('Count')

# 3. Correlation heatmap
plt.subplot(1, 2, 2)
corr_matrix = df.corr() # คำนวณค่าความสัมพันธ์
# ใช้ annot=True เพื่อแสดงตัวเลขในกล่อง และ fmt=".2f" เพื่อปัดทศนิยม 2 ตำแหน่ง
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", 
            linewidths=0.5, annot_kws={"size": 8})
plt.title('Correlation Heatmap', fontweight='bold')

plt.tight_layout() # จัดระเบียบไม่ให้กราฟซ้อนทับกัน
plt.show()
```
